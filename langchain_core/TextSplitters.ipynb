## 1. Character Text Splitter
**Important Parameters:**
- `chunk_size`: Maximum characters per chunk
- `chunk_overlap`: Characters to overlap between chunks (prevents losing context)
- `separator`: Character to split on

```python
from langchain_text_splitters import CharacterTextSplitter

text = """Albert Einstein was a German-born physicist who developed the theory of relativity. His work is also known for its influence on the philosophy of science. 
He was awarded the 1921 Nobel Prize in Physics for his services to theoretical physics,
 and especially for his discovery of the law of the photoelectric effect. This was a pivotal moment in the development of quantum theory"""
splitter = CharacterTextSplitter(separator=" ", chunk_size=100, chunk_overlap=10)
chunks = splitter.split_text(text)
for i, chunk in enumerate(chunks):
    print(f"Chunk {i+1}:\n{chunk}\n")
```

## 2. Recursive Character Text Splitter
**Key Point:** Tries separators in order - best for maintaining paragraphs and sentences.

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text = """Artificial Intelligence (AI) refers to the simulation of human intelligence by machines.\n\nIt involves machine learning, natural language processing, and neural networks.\n\nThese systems need data to function"""

splitter = RecursiveCharacterTextSplitter(
    separators=["\n\n", "\n", " ", ""], chunk_size=100, chunk_overlap=0
)
chunks = splitter.split_text(text)
for i, chunk in enumerate(chunks):
    print(f"Chunk {i+1}:\n{chunk}\n")
```

## 3. Token Text Splitter
**Important:** Uses tokenization - better for controlling actual LLM input size.

```python
from langchain_text_splitters import TokenTextSplitter

text = """Artificial Intelligence (AI) refers to the simulation of human intelligence by machines.\n\nIt involves machine learning, natural language processing, and neural networks.\n\nThese systems need data to function"""

splitter = TokenTextSplitter(chunk_size=20, chunk_overlap=5)
chunks = splitter.split_text(text)
for i, chunk in enumerate(chunks):
    print(f"Chunk {i+1}:\n{chunk}\n")
```

## 4. Language-Specific Splitters
**Key Point:** Preserves metadata for source tracking.

```python
from langchain_text_splitters import Language, RecursiveCharacterTextSplitter

python_code = """def fibonacci(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    elif n == 2:
        return [0, 1]
    else:
        fib_sequence = [0, 1]
        for i in range(2, n):
            next_fib = fib_sequence[i-1] + fib_sequence[i-2]
            fib_sequence.append(next_fib)
        return fib_sequence
print(fibonacci(10))
"""
splitter = RecursiveCharacterTextSplitter.from_language(
    language=Language.PYTHON, chunk_size=150, chunk_overlap=20
)

chunks = splitter.split_text(python_code)
for i, chunk in enumerate(chunks):
    print(f"Chunk {i+1}:\n{chunk}\n")
```
